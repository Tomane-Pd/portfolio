### Compactar uma pasta em ZIP com Python

#### 💡 Sabias que podes compactar uma pasta inteira usando apenas algumas linhas de Python?

#### Neste code, vou ensinar a criar uma pasta de exemplo, transformá-la num ficheiro ZIP e localizar o resultado. 

#### 1. Importar as bibliotecas

```python
from pathlib import Path
import shutil
```

#### 2. Criar uma pasta de exemplo

```python
pasta = Path("Meus_Ficheiros")
pasta.mkdir(exist_ok=True)

(pasta / "notas.txt").write_text(
    "Aprender Python pode poupar tempo.",
    encoding="utf-8"
)

(pasta / "tarefas.txt").write_text(
    "Organizar, compactar e partilhar.",
    encoding="utf-8"
)
```

#### 3. Compactar a pasta

```python
# Para usar uma pasta tua, altera este caminho:
# pasta = Path(r"C:\Users\Tomane Asus\Documents\Relatorios")

pasta = pasta.resolve()

if not pasta.is_dir():
    raise FileNotFoundError(f"Pasta não encontrada: {pasta}")

ficheiro_zip = shutil.make_archive(
    base_name=str(pasta.parent / pasta.name),
    format="zip",
    root_dir=pasta.parent,
    base_dir=pasta.name
)

print("Pasta compactada com sucesso!")
```
