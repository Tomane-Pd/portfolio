#### 💡 Sabias que podes baixar vídeos do YouTube usando apenas algumas linhas de Python?

##### Se precisas guardar vídeos para assistir mais tarde, o Python pode fazer isso por ti, de forma simples e rapida. 
Neste tutorial vais aprender a utilizar a biblioteca
<code>yt-dlp</code> para descarregar vídeos em alta qualidade,
de forma simples e totalmente automatizada.
</p>

1. Instale a biblioteca

```python
pip install yt-dlp -q
```

2. Importar a biblioteca

```python
from yt_dlp import YoutubeDL
```

3. Baixar o video

```python
url = "https://www.youtube.com/watch?v=Peoc886Dxnk&list=RDGMEMQ1dJ7wXfLlqCjwV0xfSNbAVMk2C5TjS2sh4&index=3" #coloque o link do video

opcoes = {
    "format": "bestvideo+bestaudio/best",
    "outtmpl": "%(title)s.%(ext)s",
    "merge_output_format": "mp4",   # Melhor qualidade possível
    "noplaylist": True              # Baixa apenas o vídeo, entao se quiser baixar toda playlist, essa parte tem que ser assim  "noplaylist": False
}

print(" A descarregar o vídeo...")

with YoutubeDL(opcoes) as ydl:
    ydl.download([url])

print("Download Feitoooooooo!")
```

4. Localizar o video Baixado

```python
import os

print(f"Vídeo foi salvo em: {os.getcwd()}")
```
