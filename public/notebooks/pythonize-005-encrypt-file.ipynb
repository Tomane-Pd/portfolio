<p style="font-size:16px;line-height:1.6;color:#555;margin:0;">
💡 <b>Sabias que podes encriptar qualquer ficheiro com uma palavra-passe utilizando Python?</b><br><br>
Ideal para proteger documentos confidenciais, bases de dados, relatórios e qualquer outro ficheiro importante.
</p>

1. Instala a biblioteca

```python
pip install cryptography
```

2. Importar os pacotes

```python
from cryptography.fernet import Fernet
import os
```

3. Proteger o arquivo

```python
# Ficheiro que será protegido
FILE = r"C:\Users\Tomane Asus\Documento.pdf"

# Nome do ficheiro encriptado
OUTPUT = r"C:\Users\Tomane Asus\Documento_Protegido.enc"

key = Fernet.generate_key() # gerar a chave

# Guardar a chave
with open("chave.key", "wb") as chave:
    chave.write(key)

fernet = Fernet(key)

with open(FILE, "rb") as ficheiro:
    dados = ficheiro.read()

dados_encriptados = fernet.encrypt(dados)

with open(OUTPUT, "wb") as ficheiro:
    ficheiro.write(dados_encriptados)

print("Ficheiro protegido com sucesso!")
print(f"Local: {os.path.abspath(OUTPUT)}")
print("Chave guardada em: chave.key")
```
