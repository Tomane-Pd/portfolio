```python
import os
from yt_dlp import YoutubeDL

# Link da playlist
playlist_url = "https://www.youtube.com/watch?v=I-9_rc-U9WE&list=PLVWgqq1nv7n5xjffNVrQRpfldCjI1pHmt"

# Pasta onde os vídeos serão salvos
download_folder = r"C:\Users\Tomane Asus\Desktop\Frances Videos"

os.makedirs(download_folder, exist_ok=True)

ydl_opts = {
    # Melhor qualidade disponível
    "format": "bv*+ba/b",

    # Salvar em MP4 quando possível
    "merge_output_format": "mp4",

    # Nome dos arquivos
    "outtmpl": os.path.join(
        download_folder,
        "%(playlist_index)03d - %(title)s.%(ext)s"
    ),

    # Continuar downloads interrompidos
    "continuedl": True,

    # Não baixar vídeos já baixados
    "download_archive": os.path.join(download_folder, "downloaded.txt"),

    # Ignorar vídeos indisponíveis
    "ignoreerrors": True,

    # Tentar novamente em caso de erro
    "retries": 10,

    # Mostrar progresso
    "quiet": False,
}

with YoutubeDL(ydl_opts) as ydl:
    ydl.download([playlist_url])

print("✅ Todos os vídeos foram baixados!")
```
