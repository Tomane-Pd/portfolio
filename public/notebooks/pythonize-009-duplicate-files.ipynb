## Encontrar ficheiros duplicados com Python

### 💡 Sabias que podes encontrar ficheiros duplicados mesmo quando têm nomes diferentes?

#### Vamos comparar o conteúdo dos ficheiros de uma pasta. O código apenas mostra os duplicados; não apaga nenhum ficheiro.

#### 1. Importar as bibliotecas

```python
from pathlib import Path
from collections import defaultdict
import hashlib
```

#### 2. Criar uma pasta de exemplo

```python
pasta = Path("Exemplo_Duplicados")
pasta.mkdir(exist_ok=True)

(pasta / "relatorio_original.txt").write_text(
    "Dados do relatório.", encoding="utf-8"
)

(pasta / "copia_relatorio.txt").write_text(
    "Dados do relatório.", encoding="utf-8"
)

(pasta / "outro_relatorio.txt").write_text(
    "Dados diferentes.", encoding="utf-8"
)
```

#### 3. Procurar os duplicados

```python
# Para usar uma pasta tua, altera este caminho:
# pasta = Path(r"C:\Users\Tomane Asus\Documents")

pasta = pasta.resolve()

if not pasta.is_dir():
    raise FileNotFoundError(f"Pasta não encontrada: {pasta}")

def calcular_hash(caminho):
    resumo = hashlib.sha256()

    with caminho.open("rb") as ficheiro:
        for bloco in iter(lambda: ficheiro.read(1024 * 1024), b""):
            resumo.update(bloco)

    return resumo.hexdigest()

grupos = defaultdict(list)

for ficheiro in pasta.rglob("*"):
    if ficheiro.is_file():
        grupos[calcular_hash(ficheiro)].append(ficheiro)
```

#### 4. Mostrar os ficheiros duplicados

```python
encontrados = False

for ficheiros in grupos.values():
    if len(ficheiros) > 1:
        encontrados = True
        print("\nFicheiros com o mesmo conteúdo:")

        for ficheiro in ficheiros:
            print(f"- {ficheiro}")

if not encontrados:
    print("Não foram encontrados ficheiros duplicados.")
```
