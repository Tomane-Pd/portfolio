💡 <b>Sabias que podes criar um QR Code totalmente personalizado utilizando Python?</b><br>
Converte qualquer texto, link, e-mail, contacto ou rede Wi-Fi num QR Code em apenas alguns segundos.
</p>

1. Instale a biblioteca

```python
pip install qrcode[pil]
```

2. Importar as bibliotecas

```python
import qrcode
import os
from IPython.display import Image, display
```

3. Criar o QR Code de um site (Lembrando que tirando o # automaticamente descomenta ou habilita ler o codigo).

```python
dados = "https://tomane-portfolio.com" # para Website ou qualquer link.

# dados = "tel:+258840000000" # Para Telefone

#dados = "mailto:teuemail@gmail.com?subject=Olá%20Fulano" # Para Email com Assunto

# dados = "https://wa.me/258840000000" # Para Contactar no WhatsApp

qr = qrcode.QRCode(
    version=1,
    error_correction=qrcode.constants.ERROR_CORRECT_H,
    box_size=12,
    border=4
)

qr.add_data(dados)
qr.make(fit=True)

imagem = qr.make_image(
    fill_color="black",
    back_color="white"
)
arquivo = "QRCode_Portfolio.png"

imagem.save(arquivo)
print("QR Code criado com sucesso!")
print(f"Local do arquivo:\n{os.path.abspath(arquivo)}")

display(Image(filename=arquivo))
```
