```python
import os
from yt_dlp import YoutubeDL

# Link da playlist
playlist_url = "https://www.youtube.com/watch?v=I-9_rc-U9WE&list=PLVWgqq1nv7n5xjffNVrQRpfldCjI1pHmt"

# Pasta de destino
download_folder = r"C:\Users\Tomane Asus\Desktop\Frances Videos"
os.makedirs(download_folder, exist_ok=True)

ydl_opts = {
    # Melhor vídeo + melhor áudio disponíveis
    "format": "bestvideo*+bestaudio/best",

    # Salvar em MP4
    "merge_output_format": "mp4",

    # Nome dos arquivos
    "outtmpl": os.path.join(
        download_folder,
        "%(playlist_index)03d - %(title)s.%(ext)s"
    ),

    # Download da playlist inteira
    "noplaylist": False,

    # Continuar downloads interrompidos
    "continuedl": True,

    # Não baixar novamente vídeos já baixados
    "download_archive": os.path.join(download_folder, "downloaded.txt"),

    # Ignorar vídeos indisponíveis
    "ignoreerrors": True,

    # Tentativas em caso de erro
    "retries": 10,

    # Mostrar progresso
    "quiet": False,

    # Se tiver FFmpeg instalado no PATH, não precisa desta linha.
    # Caso contrário, informe o caminho:
    # "ffmpeg_location": r"C:\ffmpeg\bin",
}

with YoutubeDL(ydl_opts) as ydl:
    ydl.download([playlist_url])

print("✅ Download concluído!")
```
