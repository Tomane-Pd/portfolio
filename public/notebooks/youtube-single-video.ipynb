```python
import yt_dlp

url = "https://www.youtube.com/live/1m9u10eJ3Mk?si=YzBQtW8xv06SlRPl"

opcoes = {
    'format': 'best',
    'outtmpl': '%(title)s.%(ext)s'
}

with yt_dlp.YoutubeDL(opcoes) as ydl:
    ydl.download([url])

print("Download concluído!")
```
