2. Importar as bibliotecas

```python
from rembg import remove
from PIL import Image
import os
```

3. Remover o background da imagem

```python
# Nome da imagem original
INPUT_IMAGE = r"C:\Users\Tomane Asus\Foto.jpeg"

# Nome final
OUTPUT_IMAGE = r"C:\Users\Tomane Asus\Foto_sem_fundo.png"


# Processamento

try:
    print("A carregar a imagem...")

    input_image = Image.open(INPUT_IMAGE)

    print("A remover o fundo...")

    output_image = remove(input_image)

    output_image.save(OUTPUT_IMAGE)

    print("Fundo removido com sucesso!")

    print(f"Imagem salva em:\n{os.path.abspath(OUTPUT_IMAGE)}")

except FileNotFoundError:
    print(f"A imagem '{INPUT_IMAGE}' não foi encontrada.")

except Exception as error:
    print("Ocorreu um erro:")
    print(error)
```
