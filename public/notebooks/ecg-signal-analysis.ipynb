## Processamento e Análise de Sinais de Eletrocardiograma (ECG) com Python

### Importar as Bibliotecas

```python
import pandas as pd
import numpy as np
import scipy
import matplotlib.pyplot as plt
```

### Trazer os dados

```python
ecg = scipy.datasets.electrocardiogram()
```

```python
ecg
```

### Plotar os dados

```python
# Primeiro vamos converter os dados de array para o formato pandas

ecg =pd.Series(ecg)
```

```python
ecg.plot(color='orange')
```

##### Podemos fazer um zoom dos batimentos cardiacos.

```python
ecg[:720].plot(color='orange')
```
