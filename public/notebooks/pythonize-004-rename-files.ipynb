<b>Renomeie centenas de arquivos automaticamente com Python</b>
</p>

<p style="font-size:16px;line-height:1.6;color:#555;margin:0;">
💡 <b>Sabias que podes renomear centenas de arquivos em apenas alguns segundos utilizando Python?</b><br><br>

Ideal para organizar fotografias, documentos, datasets, imagens e qualquer colecção de arquivos sem precisar renomeá-los manualmente.
</p>

1.Não é necessário instalar nenhuma biblioteca.

2. Importar o pacote

```python
import os
```

3. Organizar arquivos

```python
# Pasta contendo os arquivos
PASTA = r"C:\Users\Tomane Asus\Desktop\Pythonizar"

# Prefixo desejado
PREFIXO = "Pythonize"

# Extensão dos arquivos
EXTENSAO = ".PDF"

contador = 1

for arquivo in os.listdir(PASTA):

    if arquivo.lower().endswith(EXTENSAO):

        antigo = os.path.join(PASTA, arquivo)

        novo_nome = f"{PREFIXO}_{contador:03d}{EXTENSAO}"

        novo = os.path.join(PASTA, novo_nome)

        os.rename(antigo, novo)

        print(f" {arquivo}  →  {novo_nome}")

        contador += 1

print("Processo concluído!")
```
