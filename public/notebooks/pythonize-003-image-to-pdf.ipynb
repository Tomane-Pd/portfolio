</div>

1. Instale a biblioteca

```python
pip install pillow
```

2. Importar as bibliotecas

```python
from PIL import Image
import os
```

```python
3. Converter imagem
```

```python
# Caminho da imagem
INPUT_IMAGE = r"C:\Users\Tomane Asus\Foto.jpeg"

# Nome do PDF
OUTPUT_PDF = r"C:\Users\Tomane Asus\Foto.pdf"

try:
    print("Carregando imagem...")

    image = Image.open(INPUT_IMAGE)

    # Converter para RGB (necessário para PDF)
    if image.mode == "RGBA":
        image = image.convert("RGB")
    elif image.mode != "RGB":
        image = image.convert("RGB")

    print("Convertendo para PDF...")

    image.save(OUTPUT_PDF)

    print("Conversão concluída com sucesso!")
    print(f"Arquivo salvo em:\n{os.path.abspath(OUTPUT_PDF)}")

except FileNotFoundError:
    print("A imagem não foi encontrada.")

except Exception as error:
    print("Ocorreu um erro:")
    print(error)
```
