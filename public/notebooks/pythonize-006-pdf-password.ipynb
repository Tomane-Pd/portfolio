💡 <b>Sabias que podes proteger um PDF com uma palavra-passe utilizando Python?</b><br>
Assim, sempre que alguém tentar abrir o PDF, será necessário informar a senha.
</p>

1.  Instala a biblioteca

```python
pip install pypdf
```

2. Importar os pacotes

```python
from pypdf import PdfReader, PdfWriter
import os
```

3. Proteger um PDF

```python
# PDF original
INPUT_PDF = r"C:\Users\Tomane Asus\Documento.pdf"

# PDF protegido
OUTPUT_PDF = r"C:\Users\Tomane Asus\Documento_Protegido.pdf"

# Palavra-passe
PASSWORD = input("Crie uma palavra-passe para o PDF: ")

reader = PdfReader(INPUT_PDF)
writer = PdfWriter()

for page in reader.pages:
    writer.add_page(page)

# Proteger PDF
writer.encrypt(PASSWORD)

with open(OUTPUT_PDF, "wb") as pdf:
    writer.write(pdf)

print("PDF protegido com sucesso!")
print(f"Local: {os.path.abspath(OUTPUT_PDF)}")
```
